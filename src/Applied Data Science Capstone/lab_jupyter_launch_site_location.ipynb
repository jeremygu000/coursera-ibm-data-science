Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
def add_distance_marker(site_row, poi_lat, poi_lon, poi_name, site_map):
    """
    Add a marker and line from a launch site to a point of interest (POI).
    
    Parameters:
    - site_row: one row of launch_sites_df with 'Launch Site', 'Lat', 'Long'
    - poi_lat, poi_lon: coordinates of the point of interest
    - poi_name: name of the point of interest (e.g., 'Nearest Highway')
    - site_map: folium.Map object to update
    
    Returns:
    - site_map with marker and line added
    """
    # Extract launch site info
    launch_site_name = site_row['Launch Site']
    launch_site_lat = float(site_row['Lat'])
    launch_site_lon = float(site_row['Long'])

    # Calculate distance (km) between launch site and POI
    distance_km = calculate_distance(
        launch_site_lat, launch_site_lon, poi_lat, poi_lon
    )

    # Add marker at POI with distance in popup/label
    distance_marker = folium.Marker(
        location=[poi_lat, poi_lon],
        popup=folium.Popup(f"{poi_name}<br>{distance_km:.2f} km", max_width=250),
        icon=folium.DivIcon(
            icon_size=(20, 20), icon_anchor=(0, 0),
            html=f'<div style="font-size:12px;color:#d35400;"><b>{poi_name}: {distance_km:.2f} km</b></div>'
        )
    )
    site_map.add_child(distance_marker)

    # Draw a line from the launch site to the POI
    folium.PolyLine(
        locations=[[launch_site_lat, launch_site_lon], [poi_lat, poi_lon]],
        color='#1f77b4', weight=2
    ).add_to(site_map)

    return site_map
```

```python
def add_line_with_distance(lat1, lon1, lat2, lon2, site_map):
    """
    Draw a line between (lat1, lon1) and (lat2, lon2),
    and add a text label at the target point (lat2, lon2) showing the distance in km.
    """
    # compute distance in km
    distance_km = calculate_distance(lat1, lon1, lat2, lon2)

    # draw polyline
    line = folium.PolyLine(
        locations=[[lat1, lon1], [lat2, lon2]],
        color='#1f77b4', weight=2
    )
    site_map.add_child(line)

    # add distance marker at target point (lat2, lon2)
    distance_label = folium.map.Marker(
        [lat2, lon2],
        icon=folium.DivIcon(
            icon_size=(150, 20),
            icon_anchor=(0, -10),  # shift label slightly above the point
            html=f'<div style="font-size:12px; color:#d35400;"><b>{distance_km:.2f} km</b></div>'
        )
    )
    site_map.add_child(distance_label)

    return site_map
```
